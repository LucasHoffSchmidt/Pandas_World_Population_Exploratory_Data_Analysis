# Pandas project - Exploratory Data Analysis

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Converts csv file to a dataframe
df = pd.read_csv("world_population.csv")
df.head()
```

```python
df.shape
```

```python
# Displays each float in the dataframe with two decimals
pd.set_option("display.float_format", lambda x: "%.2f" % x)
```

```python
# Shows columns non-null count and data type
df.info()
```

```python
# Shows statistical elements for each column
df.describe()
```

```python
# Shows how many null values each column has
df.isnull().sum()
```

```python
# Shows number of unique values in each column
df.nunique()
```

```python
# Shows a list of all column names
df.columns
```

```python
# Selects columns by data type
df.select_dtypes(include="number")
```

```python
# Shows top 10 lowest values for the 2022 population
df.sort_values(by="2022 Population").head(10)
```

```python
# Shows top 10 highest populations in 2022
df.sort_values(by="2022 Population", ascending=False).head(10)
```

```python
# Shows the numerical correlations between columns
df_numeric = df.select_dtypes(include="number")
df_numeric.corr()
```

```python
# Sets figure size
plt.figure(figsize=(10, 3))

# Makes a correlation heatmap, showing the correlation numbers
sns.heatmap(df_numeric.corr(), annot = True)

# Formats the plot
plt.title("Heatmap of Numerical Correlations", fontsize=20)
```

```python
# Shows the mean of each continent, sorted by the 2022 Population, descendingly
df.groupby("Continent").mean(numeric_only=True).sort_values(by="2022 Population", ascending=False)
```

```python
# Shows all countries in Oceania
df[df["Continent"].str.contains("Oceania")]
```

```python
# Shows the mean of each continent, sorted by the 2022 Population descendingly and only including population data
df2 = df.groupby("Continent")[df.columns[5:13][::-1]].mean(numeric_only=True).sort_values(by="2022 Population", ascending=False)
df2
```

```python
# Transposes the axes to show continents in the columns and populations in the rows
df3 = df2.transpose()
df3
```

```python
# Shows the mean population by continent over time
df3.plot(figsize=(15, 6))

# Formats the plot
plt.title("Population growth over time by continent", fontsize=20, pad=20)
plt.xticks(fontsize=15, rotation=45)
plt.ylabel("Average Population Count", fontsize=15)
plt.yticks([0, 0.2e8, 0.4e8, 0.6e8, 0.8e8, 1.0e8], 
           ["0 million", "20 million", "40 million", "60 million", "80 million", "100 million"], fontsize=15)
plt.legend(fontsize=12);
```

```python
# Shows distribution of values for population over time
df[df.columns[5:13]].boxplot(figsize=(15, 6))

# Formats the plot
plt.title("Distribution of Population Count by Year", fontsize=20, pad=20)
plt.xticks(fontsize=15, rotation=45)
plt.ylabel("Population Count", fontsize=15);
```
